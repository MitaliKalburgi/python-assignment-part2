**Task 1 - Explore the Menu**

```python
# Print the full menu grouped by category

categories = ["Starters", "Mains", "Desserts"]

# print the grouped menu
for category in categories:
  print(f"\n==== {category} ====")

  for item, details in menu.items():
    if details["category"] == category:
      price = details["price"]
      available = details["available"]

      if available:                          # Availability text
        status = "Available"
      else:
        status = "Unavailable"

      print(f"{item:<20} ₹{price:<8.2f} [{status}]")
```

```python
# Using dictionary methods, to compute and print

# 1. Total number of items on the menu
total_items = len(menu)

# 2. Total number of available items
available_count = 0
for item in menu:
    if menu[item]["available"]:
        available_count += 1

# 3. The most expensive item (name + price)
max_price = 0
max_item = ""

for item, details in menu.items():
    if details["price"] > max_price:
        max_price = details["price"]
        max_item = item

# 4. All items priced under ₹150 (name + price)
print("\nItems under ₹150:-")
for item, details in menu.items():
    if details["price"] < 150:
        print(f"{item} - ₹{details['price']}")

# Final summary
print("\nSummary:")
print(f"Total items on menu      : {total_items}")
print(f"Available items          : {available_count}")
print(f"Most expensive item      : {max_item} - ₹{max_price}")
```

**Task 2 - Cart Operations**

```python
cart = []
```

```python
# Write logic to add an item to the cart

def add_to_cart(item_name, quantity):

    # Check if item exists in menu
    if item_name not in menu:
        print(f"{item_name} not found in menu")
        return

    # Check if item is marked as available
    if not menu[item_name]["available"]:
        print(f"{item_name} is unavailable")
        return

    # Check if item is already in cart
    for item in cart:
        if item["item"] == item_name:
            item["quantity"] += quantity
            return

    # Add new item to cart
    cart.append({
        "item": item_name,
        "quantity": quantity,
        "price": menu[item_name]["price"]
    })
```

```python
# Write logic to remove an item from the cart by name.

def remove_from_cart(item_name):

    # Loop through cart
    for item in cart:
        if item["item"] == item_name:
            cart.remove(item)
            return

    # If item not found
    print(f"{item_name} not in cart")
```

```python
# Write logic to update the quantity of an item already in the cart.

def update_quantity(item_name, quantity):

    # Loop through cart
    for item in cart:
        if item["item"] == item_name:
            item["quantity"] = quantity
            return

    # If item not found
    print(f"{item_name} not in cart")
```

```python
# Function to print the cart state
def print_cart():
    print("\nCurrent Cart:")

    if not cart:
        print("Cart is empty")
        return

    for item in cart:
        print(f"{item['item']} x{item['quantity']}")

# 1. Add "Paneer Tikka" × 2
add_to_cart("Paneer Tikka", 2)
print_cart()

# 2. Add "Gulab Jamun" × 1
add_to_cart("Gulab Jamun", 1)
print_cart()

# 3. Add "Paneer Tikka" × 1 (should update to 3)
add_to_cart("Paneer Tikka", 1)
print_cart()

# 4. Try to add "Mystery Burger"
add_to_cart("Mystery Burger", 1)
print_cart()

# 5. Try to add "Chicken Wings" (unavailable)
add_to_cart("Chicken Wings", 1)
print_cart()

# 6. Remove "Gulab Jamun"
remove_from_cart("Gulab Jamun")
print_cart()
```

```python
# Print a final Order Summary

print("\n========== Order Summary ==========")

subtotal = 0

# Loop through cart
for item in cart:
    item_total = item["quantity"] * item["price"]
    subtotal += item_total

    print(f"{item['item']:<20} x{item['quantity']}    ₹{item_total:.2f}")

print("------------------------------------")

# GST calculation (5%)
gst = subtotal * 0.05
total = subtotal + gst

print(f"Subtotal:                ₹{subtotal:.2f}")
print(f"GST (5%):                ₹{gst:.2f}")
print(f"Total Payable:           ₹{total:.2f}")

print("====================================")
```

**Task 3 - Inventory Tracker with Deep Copy**

```python
# Step 1 - Deep copy inventory
import copy

# Create deep copy before making any changes
inventory_backup = copy.deepcopy(inventory)

# Modify original inventory
inventory["Paneer Tikka"]["stock"] = 5

# Print both to compare
print("After modifying original inventory:\n")

print("Original Inventory (Paneer Tikka):")
print(inventory["Paneer Tikka"])

print("\nBackup Inventory (Paneer Tikka):")
print(inventory_backup["Paneer Tikka"])

# Restore original inventory
inventory = copy.deepcopy(inventory_backup)

print("\nInventory restored to original state:")
print(inventory["Paneer Tikka"])
```

```python
# Step 2 - Order Fulfilment

for item in cart:
    item_name = item["item"]
    quantity_needed = item["quantity"]
    stock_available = inventory[item_name]["stock"]

    # Case 1: Enough stock
    if stock_available >= quantity_needed:
        inventory[item_name]["stock"] -= quantity_needed

    # Case 2: Not enough stock
    else:
        print(f"⚠ Only {stock_available} units of {item_name} available, cannot fulfill full order")

        # Deduct whatever is available
        inventory[item_name]["stock"] = 0
```

```python
# Step 3 - Reorder Alert

for item_name, details in inventory.items():

    stock = details["stock"]
    reorder_level = details["reorder_level"]

    if stock <= reorder_level:
        print(f"⚠ Reorder Alert: {item_name} —  Only {stock} unit(s) left (reorder level: {reorder_level})")
```

```python
# Step 4 - Compare Inventories

print("\nCurrent Inventory (After Updates):")
for item, details in inventory.items():
    print(f"{item}: {details}")

print("\nBackup Inventory (Original):")
for item, details in inventory_backup.items():
    print(f"{item}: {details}")
```

**Task 4 - Daily Sales Log Analysis**

```python
# Step 1 - Revenue per day

print("Revenue per day:\n")

for date, orders in sales_log.items():
    total_revenue = 0

    # Loop through orders of that day
    for order in orders:
        total_revenue += order["total"]

    print(f"{date} : ₹{total_revenue:.2f}")
```

```python
# Step 2 - Print the best-selling day (date with the highest total revenue)

max_revenue = 0
best_day = ""

for date, orders in sales_log.items():
    total_revenue = 0

    # Calculate revenue for that day
    for order in orders:
        total_revenue += order["total"]

    # Check if this is the highest so far
    if total_revenue > max_revenue:
        max_revenue = total_revenue
        best_day = date

# Print result
print(f"\nBest-selling day: {best_day} — ₹{max_revenue:.2f}")
```

```python
# Step 3 - Most Ordered Item

item_count = {}

# Loop through all days
for date, orders in sales_log.items():

    # Loop through each order
    for order in orders:

        # Loop through items in that order
        for item in order["items"]:

            if item in item_count:
                item_count[item] += 1
            else:
                item_count[item] = 1

# Find most ordered item
max_count = 0
most_ordered_item = ""

for item, count in item_count.items():
    if count > max_count:
        max_count = count
        most_ordered_item = item

# Print result
print(f"\nMost ordered item: {most_ordered_item} ({max_count} orders)")
```
